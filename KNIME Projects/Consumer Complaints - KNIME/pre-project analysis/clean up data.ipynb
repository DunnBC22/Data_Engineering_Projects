# Cleaning the Input Data so Apache NiFi Can Use It

### Import Libraries

```python
import sys

import pandas as pd
```

### Retrieve Dataset

```python
df = pd.read_csv("../data/orig-complaints.csv")

df
```

### Provide Summary/Descriptive Statistics About Dataset

```python
df.describe(include='all')
```

```python
df.info()
```

### Remove Excess New Lines in String Columns

```python
for col in df.select_dtypes(include=['object', 'string']):
    df[col] = df[col].str.replace('\n', '', regex=True)
    df[col] = df[col].str.replace('( ', ' ', regex=False)
    df[col] = df[col].str.replace(' )', ' ', regex=False)
    df[col] = df[col].str.replace('  ', ' ', regex=False)
    df[col] = df[col].str.replace(' ,', ' ', regex=False)
    df[col] = df[col].str.replace(' .', ' ', regex=False)
    df[col] = df[col].str.replace(' # ', ' - ', regex=False)

df
```

### Save Cleaned Dataset

```python
df.to_csv(
    "../data/complaints.csv",
    index=False
    )
```
