## Exploratory Data Analysis with Polars 

### Load CSV File into Polars DataFrame

```python
import polars as pl
import os

def read_and_concatenate_parquet(folder_path):
    # List all Parquet files in the folder
    files = [f for f in os.listdir(folder_path) if f.endswith('.parquet')]

    df_list = []
    for file in files:
        file_path = os.path.join(folder_path, file)
        df_interim = pl.read_parquet(file_path)

        # Only cast columns that are numeric (Int64 or Float64) to Float64,
        # leaving others (e.g., strings, dates) unchanged.
        df_interim = df_interim.with_columns([
            pl.col(col).cast(pl.Float64) if df_interim.schema[col] in (pl.Int64, pl.Float64) else pl.col(col)
            for col in df_interim.columns
        ])

        df_list.append(df_interim)

    # Concatenate all DataFrames into one
    concatenated_df = pl.concat(df_list)

    return concatenated_df
```

```python
folder_path = 'data'
df = read_and_concatenate_parquet(folder_path)
df
```

### Find Longest Text Length in Each Column

```python
# Create an empty list to store max lengths for each string column
longest_text_lengths = []

# Loop through the columns to check for string columns
string_columns = [col for col in df.columns if df[col].dtype == pl.Utf8]

max_lengths = {}
for col in string_columns:
    max_length = df.select(pl.col(col).str.len_chars().max()).to_numpy()[0, 0]
    max_lengths[col] = max_length

df_max_lengths = pl.DataFrame(max_lengths)

df_max_lengths
```

### Count Unique Values in Each Column

```python
all_columns = [col for col in df.columns]

for col in all_columns:
    unique_counts = df[col].n_unique()
    print(f"Unique values in {col} :".rjust(48), f"{unique_counts}".ljust(6))
```

### Check Distribution of Numerical Columns

```python
numerical_cols = [item for item in all_columns if item not in string_columns]
numerical_cols = [item for item in numerical_cols if item not in ['id']]

numerical_cols

for col in numerical_cols:
    distribution = df.select(col).describe()
    print(col)
    print(distribution, '\n\n')
```

```python
def remove_nulls_and_count(df):
    # Drop rows with any null values
    cleaned_df = df.drop_nulls()
    # Return the number of remaining records
    return cleaned_df.height
```

```python
num_orig_records = df.height
remaining_count = remove_nulls_and_count(df)
print(f"Number of Records in the original dataframe: {num_orig_records}")
print(f"Number of remaining records: {remaining_count}")

print(f"Percent still remaining: {round(remaining_count/num_orig_records * 100, 4)}")
```
