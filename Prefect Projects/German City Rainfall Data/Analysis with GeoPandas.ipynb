## Import Dataset

```python
gdf = geopandas.read_file("data/data.csv")

gdf
```

```python
gdf.describe(include='all')
```

### Display Unique Values

```python
for column in gdf.columns:
    unique_values = gdf[column].unique()
    print(f"Column '{column}' has {len(unique_values)} unique values:")
    unique_values
```

### Number of Unique Values in Each Feature

```python
unique_counts = gdf.nunique()
print("Number of unique values per column:")
print(unique_counts)
```

### Length of Longest Character Length In Each Feature

```python
for column in gdf.select_dtypes(include='object').columns:
    max_length = gdf[column].astype(str).str.len().max()
    print(f"Longest string '{column}':\t\t{max_length}")
```
