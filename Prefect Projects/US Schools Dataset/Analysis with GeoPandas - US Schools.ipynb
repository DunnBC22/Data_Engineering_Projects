### Concatenate DataFrames into One

```python
df = pd.concat([df_public, df_private])

df.info()
```

```python
pd.set_option('display.max_columns', None)

df.head()
```

```python
pd.reset_option('display.max_columns')
```

```python
for col in df.select_dtypes(include=['object', 'string']).columns.tolist():
    u_vals = df[col].unique()
    print(f"{col} ({len(u_vals)})\n{u_vals}\n")
```

### Return Information About Unqiue Values in Select Features

```python
for col in df.select_dtypes(include=['object', 'string']).columns.tolist():
    u_vals = df[col].unique()
    if len(u_vals) < 8000:
        print(f"{col} ({len(u_vals)})\n{u_vals}\n")
    else:
        print(f"{col} ({len(u_vals)})\n")
```

```python
print(df["ZIP"].unique().tolist())
```

### Return Null Count For Each Feature

```python
print("Number of nulls in each feature")
for col in df.columns:
    print(f"{col} : {df[col].isnull().sum()}")
```

### Retrieve Longest String in Each Feature

```python
# Compute max string length per column
longest_lengths = df.astype(str).applymap(len).max()

# Format result
result = longest_lengths.reset_index()
result.columns = ["Feature", "Max String Length"]

result
```
