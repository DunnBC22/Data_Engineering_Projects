```python
import os, sys, math
import polars as pl
import math
from pathlib import Path
```

```python
print(sys.version[0:6])
print(pl.__version__)
```

```python
# Path to the folder containing the yearly files
input_folder = Path('sftp/data-orig')
output_folder = Path('sftp/data')

# Ensure the output folder exists
output_folder.mkdir(parents=True, exist_ok=True)

# Loop over each file in the input folder
for file_name in os.listdir(input_folder):
    file_path = input_folder / file_name
    
    # Check if it's a file (not a subdirectory)
    if file_path.is_file():
        print(f"Processing {file_name}...")
        
        # Read the file into a Polars DataFrame (assuming CSV format for this example)
        df = pl.read_csv(file_path)
        
        # Calculate the number of rows per quarter
        rows_per_quarter = math.ceil(df.height / 4)
        
        # Split the DataFrame into 4 quarters
        for i in range(4):
            start_row = i * rows_per_quarter
            end_row = (i + 1) * rows_per_quarter if (i + 1) * rows_per_quarter < df.height else df.height
            quarter_df = df.slice(start_row, end_row - start_row)
            
            # Save the quartered DataFrame into the output folder
            output_file_name = f"{Path(file_name).stem}_quarter_{i + 1}.csv"
            quarter_df.write_csv(output_folder / output_file_name)
            print(f"  - Saved: {output_file_name}")

print("Processing complete!")
```
