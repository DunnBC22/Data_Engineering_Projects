```python
import pandas as pd
from tabulate import tabulate
from pandas.api.types import is_string_dtype

def find_and_display_longest_strings(df):
    # Initialize a list to store results
    results = []
    
    # Iterate through columns
    for column in df.columns:
        # Check if the column has string data type
        if is_string_dtype(df[column]):
            # Find the longest string in the column
            longest_string_index = df[column].astype(str).str.len().idxmax()
            longest_string = df.loc[longest_string_index, column]
            # Store the result
            results.append([column, longest_string, len(longest_string)])
    
    # Create a DataFrame from the results
    result_df = pd.DataFrame(results, columns=['Column', 'Longest String', 'Length'])
    
    # Sort the results by length in descending order
    result_df = result_df.sort_values('Length', ascending=False).reset_index(drop=True)
    
    # Display the results in a pretty table
    print("\nLongest Strings in Each Column:")
    print(tabulate(result_df, headers='keys', tablefmt='pretty', showindex=False))
    
    return result_df

result = find_and_display_longest_strings(df)
```
