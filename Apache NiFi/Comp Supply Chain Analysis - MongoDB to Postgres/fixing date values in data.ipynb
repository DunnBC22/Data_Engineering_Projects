# Fix Date Values in Dataset

```python
import sys

import pandas as pd
```

```python
sys.version[0:6]
pd.__version__
```

```python
df = pd.read_csv("mongodb_scripts/data-original.csv")

df
```

```python
df['procured_date'].unique()
```

```python
def fix_dates(date):
    parts = date.split('/')
    if len(parts[-1]) == 2:  # 2-digit year
        parts[-1] = '20' + parts[-1]
    return '/'.join(str(int(part)) for part in parts)  # Remove leading zeros

df['procured_date'] = df['procured_date'].apply(fix_dates)

df['procured_date'].unique()
```
