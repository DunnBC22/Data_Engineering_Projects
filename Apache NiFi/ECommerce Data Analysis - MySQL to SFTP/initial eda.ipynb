```python
import pandas as pd
```

### Fact Table

```python
df = pd.read_csv("mysql_scripts/fact_table.csv")

df
```

```python
df.info()
```

```python
for col in df.columns:
    unique_values = df[col].unique()
    if len(unique_values) < 101:
        print(f"unique values in {col} ({len(unique_values)}): {unique_values}")
    else:
        print(f"unique values in {col} ({len(unique_values)})")
```

```python
def longest_string_length_per_column(df):
    df = df.astype(str)
    return df.apply(lambda col: col.str.len().max())

longest_lengths = longest_string_length_per_column(df)
print(longest_lengths)
```
